```python
# =========================================================
# Geocoding with handling for geographical_region = "NR" (for agmp_data_wip)
# =========================================================
from geopy.geocoders import Nominatim, ArcGIS
from geopy.extra.rate_limiter import RateLimiter
from geopy.exc import GeocoderQueryError, GeocoderTimedOut, GeocoderServiceError
import pandas as pd
import numpy as np
import time

# =========================================================
# 1. Safety checks
# =========================================================
if 'agmp_data_wip' not in globals():
    raise NameError("agmp_data_wip is not defined. Please ensure your DataFrame is loaded.")

# Remove duplicate columns
agmp_data_wip = agmp_data_wip.loc[:, ~agmp_data_wip.columns.duplicated()]

if 'country_participant_01' not in agmp_data_wip.columns:
    raise KeyError("Column 'country_participant_01' not found in agmp_data_wip.")

# Force Series (avoids .unique() error)
participant_series = pd.Series(agmp_data_wip['country_participant_01'])

# =========================================================
# 2. Prepare address list
# =========================================================
addresses = participant_series.dropna().astype(str)
valid_addresses = [addr.strip() for addr in addresses if addr.strip() and len(addr.strip()) > 1]
valid_addresses = pd.Series(valid_addresses).unique()

print(f"Processing {len(valid_addresses)} valid addresses out of {len(addresses.unique())} total unique entries.")

# =========================================================
# 3. Initialize geocoders
# =========================================================
primary_geolocator = Nominatim(user_agent="myApp01")
primary_geocode = RateLimiter(primary_geolocator.geocode, min_delay_seconds=1.2)
fallback_geolocator = ArcGIS(scheme="https")
fallback_geocode = RateLimiter(fallback_geolocator.geocode, min_delay_seconds=1.2)

# =========================================================
# 4. Geocoding cache
# =========================================================
location_cache = {}

def geocode_with_fallback(address):
    """Try Nominatim first, then ArcGIS if it fails"""
    cleaned_address = str(address).strip()

    # Primary: Nominatim
    try:
        location = primary_geocode(cleaned_address)
        if location:
            return (location.latitude, location.longitude, "Nominatim")
    except (GeocoderQueryError, GeocoderTimedOut, GeocoderServiceError) as e:
        print(f"  Nominatim failed for '{cleaned_address}': {e}")
    except Exception as e:
        print(f"  Nominatim unexpected error for '{cleaned_address}': {e}")

    # Fallback: ArcGIS
    try:
        time.sleep(0.5)
        location = fallback_geocode(cleaned_address)
        if location:
            return (location.latitude, location.longitude, "ArcGIS")
    except (GeocoderQueryError, GeocoderTimedOut, GeocoderServiceError) as e:
        print(f"  ArcGIS failed for '{cleaned_address}': {e}")
    except Exception as e:
        print(f"  ArcGIS unexpected error for '{cleaned_address}': {e}")

    return (np.nan, np.nan, "Failed")

# =========================================================
# 5. Geocode loop
# =========================================================
start_time = time.time()
for i, address in enumerate(valid_addresses):
    lat, lon, source = geocode_with_fallback(address)
    location_cache[address] = (lat, lon)

    if pd.notna(lat):
        print(f"✓ {i+1}/{len(valid_addresses)}: {address} ({source})")
    else:
        print(f"✗ {i+1}/{len(valid_addresses)}: {address} (not found)")

    if (i + 1) % 5 == 0:
        elapsed = time.time() - start_time
        print(f"  Progress: {i+1}/{len(valid_addresses)} completed in {elapsed:.1f}s")

# =========================================================
# 6. Fill cache for invalid addresses
# =========================================================
for address in addresses.unique():
    if address not in location_cache:
        location_cache[address] = (np.nan, np.nan)

# =========================================================
# 7. Map results to DataFrame (lat first, lon second)
# =========================================================
agmp_data_wip['latitude_01'] = participant_series.map(lambda x: location_cache.get(str(x).strip(), (np.nan, np.nan))[0])
agmp_data_wip['longitude_01'] = participant_series.map(lambda x: location_cache.get(str(x).strip(), (np.nan, np.nan))[1])

# =========================================================
# 8. Handle geographical_region = "NR"
# =========================================================
nr_count = 0
if 'geographical_region' in agmp_data_wip.columns:
    nr_mask = agmp_data_wip['geographical_region'].astype(str).str.upper() == "NR"
    agmp_data_wip.loc[nr_mask, ['latitude_01', 'longitude_01']] = np.nan
    nr_count = nr_mask.sum()
    valid_mask = ~nr_mask
else:
    valid_mask = pd.Series(True, index=agmp_data_wip.index)

# =========================================================
# 9. Success rate (excluding NR)
# =========================================================
successful_geocodes = agmp_data_wip.loc[valid_mask, 'latitude_01'].notna().sum()
total_valid_rows = valid_mask.sum()

if total_valid_rows > 0:
    success_rate = (successful_geocodes / total_valid_rows) * 100
    print(f"\nGeocoding completed! Success rate: {successful_geocodes}/{total_valid_rows} ({success_rate:.1f}%)")
    if nr_count > 0:
        print(f"Excluded {nr_count} records with geographical_region = 'NR'")
else:
    print("\nGeocoding completed! No valid addresses to geocode.")

# =========================================================
# 10. Ensure numeric dtypes
# =========================================================
agmp_data_wip['latitude_01'] = pd.to_numeric(agmp_data_wip['latitude_01'], errors='coerce')
agmp_data_wip['longitude_01'] = pd.to_numeric(agmp_data_wip['longitude_01'], errors='coerce')
```
